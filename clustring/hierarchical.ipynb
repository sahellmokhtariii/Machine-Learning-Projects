```python
import pandas as pd
import numpy as np
from scipy import ndimage 
from scipy.cluster import hierarchy 
from scipy.spatial import distance_matrix 
from matplotlib import pyplot as plt 
from sklearn import manifold, datasets 
from sklearn.cluster import AgglomerativeClustering 
from sklearn.datasets import make_blobs 
```

```python
X1 , y1 = make_blobs(n_samples=50 , centers=[[4,4], [-2, -1], [1, 1], [10,4]], cluster_std=0.9)
```

```python
plt.scatter(X1[:,0] , X1[:,1] , marker='o')
```

```python
agglom = AgglomerativeClustering (n_clusters=4 , linkage="average")
```

```python
agglom.fit(X1 , y1)
```

```python
plt.Figure(figsize=(6,4))

x_min , x_max = np.min(X1 , axis=0) , np.max(X1 , axis=0)
X1 = (X1 - x_min) / (x_max - x_min)
for i in range (X1.shape[0]):
    #نوشتن متن روی مختصات مشخص
    plt.text(X1[i , 0] , X1[i,1] , str(y1[i]),
             #پالت رنگی
             color=plt.cm.nipy_spectral(agglom.labels_[i] / 10.),
             fontdict={'weight': 'bold', 'size': 9})
    
plt.xticks([])
plt.yticks([])


# رسم نقاط
plt.scatter(X1[:, 0], X1[:, 1], marker='.')
plt.show()
```

```python
dist_matrix = distance_matrix(X1, X1)
print(dist_matrix)
```

```python
z = hierarchy.linkage(dist_matrix, 'complete')
```

مقایسه روش‌های مختلف linkage:
روش	     چطور حساب میشه                 	رفتار
'complete'	دورترین نقاط	  خوشه‌های فشرده و کوچک می‌سازه
'single'	نزدیکترین   نقاط    	خوشه‌های کشیده و زنجیره‌ای
'average'	میانگین همه فاصله‌ها	           بینابین، متعادل
'ward'	کمترین افزایش واریانس      	خوشه‌های کروی و هم اندازه


```python
dendro = hierarchy.dendrogram(z)
```
