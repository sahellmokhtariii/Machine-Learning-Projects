Lets do some practice about Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.plot([1 , 2 , 3 , 4])
plt.ylabel("funny numbers")
plt.show()
```

```python
plt.plot([1, 3, 4, 9],[8, 5, -1, 3])
```

```python
plt.plot([1, 3, 4, 9],[8, 5, -1, 3], "ro")
```

```python
t = np.arange(0 , 5 , 0.2)
plt.plot(t ** 2 , "--r")
```

```python
t = np.arange(0 , 5 , 0.2)
plt.plot(t , t , "r-." , t , t**2 , "bs" , t , t**3 , "^g")
```
